```python
from pyspark.sql.functions import countDistinct

df = spark.table(FACT_TABLE)

total = df.count()

unique = df.select(countDistinct("flight_sk")).collect()[0][0]

assert total == unique

print(" Test 3 Passed - flight_sk is unique")
```

```python
from pyspark.sql.functions import col

df = spark.table(FACT_TABLE)

assert df.filter(col("airline_sk").isNull()).count() == 0
assert df.filter(col("origin_airport_sk").isNull()).count() == 0
assert df.filter(col("dest_airport_sk").isNull()).count() == 0

print(" Test 4 Passed - Foreign Keys validated")
```
